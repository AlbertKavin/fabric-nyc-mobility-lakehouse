```python
spark.sql("CREATE SCHEMA IF NOT EXISTS ops")

rows = [
    ("2024-01", "2024-01-01", "2024-01-31", 1),
    ("2024-02", "2024-02-01", "2024-02-29", 1),
    ("2024-03", "2024-03-01", "2024-03-31", 1),
]
df = spark.createDataFrame(rows, ["year_month", "start_date", "end_date", "is_active"])
df.write.mode("overwrite").format("delta").saveAsTable("ops.pipeline_config")

spark.sql("SELECT * FROM ops.pipeline_config").show()
```
